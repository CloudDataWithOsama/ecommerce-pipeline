### Gold Layer Analytics View Creation
##### **Catalog:** `ecommerce` | **Schema:** `gold` | **View:** `v_ecommerce_analytics_summary`

##### Create Analytics View using PySpark SQL

```python
spark.sql("""
CREATE OR REPLACE VIEW ecommerce.gold.gld_denormalized_view AS
SELECT 
    -- Fact Fields
    f.transaction_id,
    f.seq_no,
    f.transaction_date,
    f.transaction_ts,
    f.channel,
    f.coupon_code,
    f.coupon_flag,
    f.unit_price_currency,
    
    -- Customer Dimension Fields
    c.customer_id,
    c.phone,
    c.country_code,
    c.country,
    c.state,
    c.region,
    
    -- Product Dimension Fields
    p.product_id,
    p.sku,
    p.category_code,
    p.category_name,
    p.brand_code,
    p.brand_name,
    p.color,
    p.size,
    p.material,
    
    -- Date Dimension Fields
    d.date_id,
    d.date AS dimension_date,
    d.year,
    d.quarter,
    d.month_name,
    d.day_name,
    d.week,
    d.is_weekend,
    
    -- Metrics & Measures
    f.quantity,
    f.unit_price,
    f.gross_amount,
    f.discount_percent,
    f.discount_amount,
    f.tax_amount,
    f.net_amount,
    f.net_amount_inr

FROM ecommerce.gold.gld_fact_order_items f
LEFT JOIN ecommerce.gold.gld_dim_customers c 
    ON f.customer_id = c.customer_id
LEFT JOIN ecommerce.gold.gld_dim_products p 
    ON f.product_id = p.product_id
LEFT JOIN ecommerce.gold.gld_dim_date d 
    ON f.date_id = d.date_id
""")
```
